```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("Final project/ai4i2020.csv")
print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Air temperature [K]"], bins=20)

plt.title("Air Temperature Distribution")

plt.xlabel("Air Temperature (K)")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(6,4))

df["Machine failure"].value_counts().plot(kind="bar")

plt.title("Machine Failure Count")

plt.xlabel("Machine Failure")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.scatter(df["Rotational speed [rpm]"], df["Torque [Nm]"])

plt.title("Rotational Speed vs Torque")

plt.xlabel("Rotational Speed (rpm)")

plt.ylabel("Torque (Nm)")
```

```python
plt.figure(figsize=(8,4))

plt.plot(df["Air temperature [K]"])

plt.title("Air Temperature")

plt.xlabel("Sample Number")

plt.ylabel("Temperature (K)")

plt.show()
```

```python
plt.figure(figsize=(6,6))

df["Type"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Machine Type Distribution")

plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(10,6))

corr = df.select_dtypes(include=['int64','float64']).corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="Type", hue="Machine failure", data=df)

plt.title("Machine Failure by Type")

plt.xlabel("Machine Type")

plt.ylabel("Count")

plt.show()
```

```python
df.groupby("Type")[["Air temperature [K]",
                    "Process temperature [K]",
                    "Rotational speed [rpm]",
                    "Torque [Nm]"]].mean()
```

```python
plt.figure(figsize=(8,5))

df[['TWF','HDF','PWF','OSF','RNF']].sum().plot(kind='bar')

plt.title("Failure Types Count")

plt.xlabel("Failure Type")

plt.ylabel("Count")

plt.show()     
```
